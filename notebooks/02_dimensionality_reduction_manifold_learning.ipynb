# 02 — Dimensionality Reduction and Manifold Learning

This notebook compares linear and non-linear dimensionality reduction methods.

The goal is to understand when embeddings are useful and when they can mislead.


```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
from sklearn.decomposition import KernelPCA, PCA
from sklearn.manifold import Isomap, LocallyLinearEmbedding
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

sys.path.append(str(Path.cwd() / "src"))

from unsup_lab.data import make_customer_segmentation_data
from unsup_lab.plotting import plot_embedding
```

```python
dataset = make_customer_segmentation_data(n_customers=1_500, random_state=7)
features = dataset.features

x = StandardScaler().fit_transform(features)
```

## PCA

```python
pca = PCA(n_components=6, random_state=7)
pca.fit(x)

explained = pca.explained_variance_ratio_

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(explained) + 1), explained.cumsum(), marker="o")
plt.title("Cumulative explained variance")
plt.xlabel("Number of components")
plt.ylabel("Cumulative explained variance")
plt.tight_layout()
plt.show()
```

```python
pca_2d = PCA(n_components=2, random_state=7).fit_transform(x)
plot_embedding(pca_2d, title="PCA embedding")
```

## Non-linear embeddings

```python
methods = {
    "Kernel PCA": KernelPCA(n_components=2, kernel="rbf", gamma=0.2),
    "Isomap": Isomap(n_components=2, n_neighbors=12),
    "Locally Linear Embedding": LocallyLinearEmbedding(
        n_components=2,
        n_neighbors=12,
        random_state=7,
    ),
}

for name, method in methods.items():
    embedding = method.fit_transform(x)
    plot_embedding(embedding, title=name)
```

## Notes

Two-dimensional plots are useful for communication, but visual separation is not proof of reliable segmentation. Always compare embedding quality, downstream stability, and sensitivity to preprocessing choices.
